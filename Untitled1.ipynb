```python
def is_year_leap(year):
    """
    Возвращает True, если год високосный, иначе False.

    Високосный год:
    - делится на 4 и не делится на 100, ИЛИ
    - делится на 400.
    """
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
```

```python
def is_prime(n: int) -> bool:
    """Возвращает True, если n — простое число, иначе False."""
    # Числа меньше 2 не являются простыми
    if n < 2:
        return False

    # 2 и 3 — простые
    if n == 2 or n == 3:
        return True

    # Чётные числа и кратные 3 (кроме 2 и 3) — не простые
    if n % 2 == 0 or n % 3 == 0:
        return False

    # Проверяем делители от 5 до sqrt(n) с шагом 6
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6

    return True


# Пример использования с вводом от пользователя
try:
    n = int(input("Введите целое число: "))
    if is_prime(n):
        print(f"{n} — простое число")
    else:
        print(f"{n} — не простое число")
except ValueError:
    print("Пожалуйста, введите корректное целое число.")
```

```python
def swap_if_needed(a, b):
    # Вычисляем границы промежутка для второй переменной
    left_bound = ((-138 / 2) ** 1.3)

    # Считаем правую границу аккуратно: сначала степень, потом умножение
    inner = -69 / (28 ** 5.1)
    right_bound = abs(inner * 4)

    # Условие 1: a больше b в 3.6 раза (a > 3.6 * b)
    condition_1 = a > 3.6 * b

    # Условие 2: b находится в промежутке [left_bound, right_bound]
    condition_2 = left_bound <= b <= right_bound

    if condition_1 or condition_2:
        a, b = b, a  # меняем местами
        return a, b, True  # вернули новые значения и флаг, что обмен был
    else:
        return a, b, False  # значения не менялись


# Пример использования
a = float(input("Введите значение первой переменной (a): "))
b = float(input("Введите значение второй переменной (b): "))

new_a, new_b, swapped = swap_if_needed(a, b)

if swapped:
    print(f"Условие выполнено: поменяли местами. Теперь a = {new_a}, b = {new_b}")
else:
    print(f"Условия не выполнены: значения остались прежними. a = {a}, b = {b}")

print(f"Границы промежутка для b: [{left_bound:.6f}, {right_bound:.6e}]")
```

```python
def analyze_numbers():
    numbers = []

    # Ввод 5 чисел
    for i in range(5):
        while True:
            try:
                value = float(input(f"Введите число {i + 1}: "))
                numbers.append(value)
                break
            except ValueError:
                print("Пожалуйста, введите корректное число.")

    # Проверка на уникальность: сравниваем длину списка и множества
    are_unique = len(numbers) == len(set(numbers))

    # Подсчёт категорий
    even_count = 0          # чётные (только для целых значений)
    negative_count = 0      # отрицательные
    in_range_count = 0      # в промежутке [-256; 1024]

    for n in numbers:
        # Чётность проверяем только если число целое (например, 4.0 считаем целым)
        if n.is_integer() and int(n) % 2 == 0:
            even_count += 1

        if n < 0:
            negative_count += 1

        if -256 <= n <= 1024:
            in_range_count += 1

    # Вывод результатов
    print("\n--- Результаты ---")
    print(f"Введённые числа: {numbers}")
    print(f"Все числа уникальны: {are_unique}")
    print(f"Количество чётных чисел: {even_count}")
    print(f"Количество отрицательных чисел: {negative_count}")
    print(f"Количество чисел в промежутке [-256; 1024]: {in_range_count}")
```
